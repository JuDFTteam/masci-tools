```python
import json
```

```python
# read json file
with open('defaults_kkrhost.json') as _f:
    d = json.load(_f)
```

```python
# format:  {key: [default value, formatting string, is mandatory, docstring]} (see README.txt for details)
d['<COARSEN_KPTS>'] = [None, '%l', False, "Run option: activate grid coarsening for k-point integration (useful in scf calculations with many k-points)."]
```

```python
# write out sorted
with open('defaults_kkrhost_new.json', 'w') as _f:
    json.dump({k:d[k] for k in sorted(d.keys())}, _f, indent=2)
```
